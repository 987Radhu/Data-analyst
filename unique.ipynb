```python
import seaborn as sns
import matplotlib.pyplot as plt

# Group by gender and sum the salary
compcount = df.groupby("gender")["salary"].sum().reset_index()

# Extract data and labels
labels = compcount["gender"]
data = compcount["salary"]

# Set color palette
colorpallet = sns.color_palette("coolwarm")

# Plot the pie chart
plt.pie(data, labels=labels, colors=colorpallet, autopct="%1.2f%%")

# Show the plot
plt.show()
```

```python

compcount = df.groupby("dept")["salary"].sum().reset_index()


cols = compcount["dept"]
data = compcount["salary"]


sorted_depts = compcount.sort_values("salary", ascending=False)["dept"]

sns.barplot(x=cols, y=data, order=sorted_depts)


plt.title("Department Salary Distribution")
plt.xlabel("Department")
plt.ylabel("Salary")


plt.show()
```
